```python
%%writefile confusion_analysis.py
import json
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from sklearn.metrics import confusion_matrix, classification_report
from transformers import AutoTokenizer
from torch.utils.data import DataLoader

from src.data_splits import make_splits
from src.model import GrieveAIClassifier, load_taxonomy, MURIL_CHECKPOINT
from src.train import GrievanceDataset


# -----------------------------
# Settings
# -----------------------------
DATA = "data/processed/grievances_synthetic.csv"
TAXONOMY = "config/taxonomy.json"
CHECKPOINT = "checkpoints/muril_lora/run2"
BATCH_SIZE = 16
MAX_LENGTH = 128
SEED = 42

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)


# -----------------------------
# Load taxonomy + data
# -----------------------------
taxonomy = load_taxonomy(TAXONOMY)
df = pd.read_csv(DATA)

train_df, val_df, test_df = make_splits(df, seed=SEED)

print(
    f"Train: {len(train_df)} | "
    f"Val: {len(val_df)} | "
    f"Test: {len(test_df)}"
)

tokenizer = AutoTokenizer.from_pretrained(
    str(Path(CHECKPOINT) / "tokenizer")
)

test_ds = GrievanceDataset(
    test_df,
    tokenizer,
    taxonomy,
    MAX_LENGTH
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)


# -----------------------------
# Load Run 2 model
# -----------------------------
model = GrieveAIClassifier(
    len(taxonomy["categories"]),
    len(taxonomy["subcategories"])
)

model.to(device)

# Load LoRA adapter + heads
model.encoder.load_adapter(
    str(Path(CHECKPOINT) / "adapter"),
    adapter_name="default"
)

heads = torch.load(
    str(Path(CHECKPOINT) / "heads.pt"),
    map_location=device
)

model.category_head.load_state_dict(heads["category_head"])
model.subcategory_head.load_state_dict(heads["subcategory_head"])
model.priority_head.load_state_dict(heads["priority_head"])

model.eval()


# -----------------------------
# Collect predictions
# -----------------------------
true_cat = []
pred_cat = []

true_sub = []
pred_sub = []

true_sub_oracle = []
pred_sub_oracle = []

texts = []

with torch.no_grad():

    start_idx = 0

    for batch in test_loader:

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        token_type_ids = batch.get("token_type_ids")

        if token_type_ids is not None:
            token_type_ids = token_type_ids.to(device)

        out = model(
            input_ids,
            attention_mask,
            token_type_ids=token_type_ids
        )

        # Category prediction
        cat = out["category_logits"].argmax(-1)

        # Subcategory using predicted category
        masked_sub = torch.stack([
            model.mask_subcategory_logits(
                out["subcategory_logits"][i],
                int(cat[i]),
                taxonomy
            )
            for i in range(len(cat))
        ])

        sub = masked_sub.argmax(-1)

        # Subcategory using GOLD category
        oracle_masked_sub = torch.stack([
            model.mask_subcategory_logits(
                out["subcategory_logits"][i],
                int(batch["category_label"][i]),
                taxonomy
            )
            for i in range(len(batch["category_label"]))
        ])

        oracle_sub = oracle_masked_sub.argmax(-1)

        true_cat.extend(batch["category_label"].cpu().tolist())
        pred_cat.extend(cat.cpu().tolist())

        true_sub.extend(batch["subcategory_label"].cpu().tolist())
        pred_sub.extend(sub.cpu().tolist())

        true_sub_oracle.extend(
            batch["subcategory_label"].cpu().tolist()
        )
        pred_sub_oracle.extend(
            oracle_sub.cpu().tolist()
        )


# -----------------------------
# Names
# -----------------------------
categories = taxonomy["categories"]
subcategories = taxonomy["subcategories"]


# -----------------------------
# Category confusion matrix
# -----------------------------
cat_cm = confusion_matrix(
    true_cat,
    pred_cat,
    labels=list(range(len(categories)))
)

cat_cm_df = pd.DataFrame(
    cat_cm,
    index=categories,
    columns=categories
)

cat_cm_df.to_csv(
    "run2_category_confusion_matrix.csv"
)


# -----------------------------
# Subcategory confusion matrix
# -----------------------------
sub_cm = confusion_matrix(
    true_sub,
    pred_sub,
    labels=list(range(len(subcategories)))
)

sub_cm_df = pd.DataFrame(
    sub_cm,
    index=subcategories,
    columns=subcategories
)

sub_cm_df.to_csv(
    "run2_subcategory_confusion_matrix.csv"
)


# -----------------------------
# Oracle subcategory confusion
# -----------------------------
oracle_cm = confusion_matrix(
    true_sub_oracle,
    pred_sub_oracle,
    labels=list(range(len(subcategories)))
)

oracle_cm_df = pd.DataFrame(
    oracle_cm,
    index=subcategories,
    columns=subcategories
)

oracle_cm_df.to_csv(
    "run2_oracle_subcategory_confusion_matrix.csv"
)


# -----------------------------
# Per-class report
# -----------------------------
report = classification_report(
    true_sub,
    pred_sub,
    labels=list(range(len(subcategories))),
    target_names=subcategories,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).T

report_df.to_csv(
    "run2_subcategory_classification_report.csv"
)


# -----------------------------
# Find strongest confusion pairs
# -----------------------------
pairs = []

for i in range(len(subcategories)):
    for j in range(len(subcategories)):

        if i == j:
            continue

        count = sub_cm[i, j]

        if count > 0:
            pairs.append({
                "true_subcategory": subcategories[i],
                "predicted_subcategory": subcategories[j],
                "count": int(count)
            })

pairs_df = (
    pd.DataFrame(pairs)
    .sort_values("count", ascending=False)
)

pairs_df.to_csv(
    "run2_top_subcategory_confusions.csv",
    index=False
)


# -----------------------------
# Print useful summary
# -----------------------------
print("\n" + "=" * 70)
print("RUN 2 — CATEGORY CONFUSION MATRIX")
print("=" * 70)
print(cat_cm_df)

print("\n" + "=" * 70)
print("RUN 2 — TOP SUBCATEGORY CONFUSIONS")
print("=" * 70)
print(pairs_df.head(30).to_string(index=False))

print("\n" + "=" * 70)
print("RUN 2 — WORST SUBCATEGORIES BY F1")
print("=" * 70)

worst = (
    report_df
    .loc[subcategories, ["precision", "recall", "f1-score", "support"]]
    .sort_values("f1-score")
)

print(worst.to_string())

print("\nSaved files:")
print("run2_category_confusion_matrix.csv")
print("run2_subcategory_confusion_matrix.csv")
print("run2_oracle_subcategory_confusion_matrix.csv")
print("run2_subcategory_classification_report.csv")
print("run2_top_subcategory_confusions.csv")
```

```python
%%writefile hybrid_eval.py

import json
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from sklearn.metrics import f1_score, mean_absolute_error, recall_score
from transformers import AutoTokenizer
from torch.utils.data import DataLoader

from src.data_splits import make_splits
from src.model import GrieveAIClassifier, load_taxonomy
from src.train import GrievanceDataset
from src.baseline_classifier import BaselineGrievanceClassifier


DATA = "data/processed/grievances_synthetic.csv"
TAXONOMY = "config/taxonomy.json"
CHECKPOINT = "checkpoints/muril_lora/run2"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

# -----------------------------
# Load exact same split
# -----------------------------
taxonomy = load_taxonomy(TAXONOMY)
df = pd.read_csv(DATA)

train_df, val_df, test_df = make_splits(df, seed=42)

print(
    f"Train: {len(train_df)} | "
    f"Val: {len(val_df)} | "
    f"Test: {len(test_df)}"
)

# -----------------------------
# Load MuRIL Run 2
# -----------------------------
tokenizer = AutoTokenizer.from_pretrained(
    str(Path(CHECKPOINT) / "tokenizer")
)

test_ds = GrievanceDataset(
    test_df,
    tokenizer,
    taxonomy,
    max_length=128
)

test_loader = DataLoader(
    test_ds,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

model = GrieveAIClassifier(
    len(taxonomy["categories"]),
    len(taxonomy["subcategories"])
)

model.to(device)

model.encoder.load_adapter(
    str(Path(CHECKPOINT) / "adapter"),
    adapter_name="default"
)

heads = torch.load(
    str(Path(CHECKPOINT) / "heads.pt"),
    map_location=device
)

model.category_head.load_state_dict(heads["category_head"])
model.subcategory_head.load_state_dict(heads["subcategory_head"])
model.priority_head.load_state_dict(heads["priority_head"])

model.eval()

# -----------------------------
# Load baseline
# -----------------------------
baseline = BaselineGrievanceClassifier.load(
    path="checkpoints/baseline",
    taxonomy_path=TAXONOMY
)

print("Loaded TF-IDF + Logistic Regression baseline.")


# -----------------------------
# Collect MuRIL categories
# -----------------------------
muril_categories = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        token_type_ids = batch.get("token_type_ids")

        if token_type_ids is not None:
            token_type_ids = token_type_ids.to(device)

        out = model(
            input_ids,
            attention_mask,
            token_type_ids=token_type_ids
        )

        cat = out["category_logits"].argmax(-1)

        muril_categories.extend(cat.cpu().tolist())


# -----------------------------
# Category names
# -----------------------------
categories = taxonomy["categories"]
subcategories = taxonomy["subcategories"]

muril_category_names = [
    categories[i] for i in muril_categories
]


# -----------------------------
# Hybrid subcategory prediction
# -----------------------------
hybrid_subcategories = []

for text, predicted_category in zip(
    test_df["text"].tolist(),
    muril_category_names
):

    X = baseline.vectorizer.transform([text])

    sub_probs = baseline.subcategory_clf.predict_proba(X)[0]
    sub_classes = list(baseline.subcategory_clf.classes_)

    valid_subs = taxonomy["categories"][predicted_category]["subcategories"]

    candidates = [
        (label, float(prob))
        for label, prob in zip(sub_classes, sub_probs)
        if label in valid_subs
    ]

    predicted_subcategory = max(
        candidates,
        key=lambda x: x[1]
    )[0]

    hybrid_subcategories.append(predicted_subcategory)


# -----------------------------
# Ground truth
# -----------------------------
true_categories = test_df["category"].tolist()
true_subcategories = test_df["subcategory"].tolist()


# -----------------------------
# Metrics
# -----------------------------
category_f1 = f1_score(
    true_categories,
    muril_category_names,
    average="macro"
)

subcategory_f1 = f1_score(
    true_subcategories,
    hybrid_subcategories,
    average="macro"
)


# -----------------------------
# Print results
# -----------------------------
print("\n" + "=" * 70)
print("HYBRID RESULTS")
print("=" * 70)

print(f"MuRIL category Macro-F1:       {category_f1:.4f}")
print(f"TF-IDF/LogReg subcategory F1:  {subcategory_f1:.4f}")

print("\nReference:")
print("MuRIL Run 2 category F1:       0.9859")
print("MuRIL Run 2 subcategory F1:     0.3046")
print("TF-IDF/LogReg category F1:      0.8641")
print("TF-IDF/LogReg subcategory F1:   0.6954")


# -----------------------------
# Save predictions
# -----------------------------
results = pd.DataFrame({
    "text": test_df["text"].tolist(),
    "true_category": true_categories,
    "muril_category": muril_category_names,
    "true_subcategory": true_subcategories,
    "hybrid_subcategory": hybrid_subcategories
})

results.to_csv(
    "run4_hybrid_predictions.csv",
    index=False
)

print("\nSaved: run4_hybrid_predictions.csv")
```

```python
%%writefile hybrid_eval.py

import json
from pathlib import Path

import pandas as pd
import torch
from sklearn.metrics import f1_score
from transformers import AutoTokenizer
from torch.utils.data import DataLoader

from src.data_splits import make_splits
from src.model import GrieveAIClassifier, load_taxonomy
from src.train import GrievanceDataset
from src.baseline_classifier import BaselineGrievanceClassifier


DATA = "data/processed/grievances_synthetic.csv"
TAXONOMY = "config/taxonomy.json"
CHECKPOINT = "checkpoints/muril_lora/run2"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)


# ============================================================
# Load data and exact same split
# ============================================================

taxonomy = load_taxonomy(TAXONOMY)
df = pd.read_csv(DATA)

train_df, val_df, test_df = make_splits(df, seed=42)

print(
    f"Train: {len(train_df)} | "
    f"Val: {len(val_df)} | "
    f"Test: {len(test_df)}"
)


# ============================================================
# Load MuRIL Run 2
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    str(Path(CHECKPOINT) / "tokenizer")
)

test_ds = GrievanceDataset(
    test_df,
    tokenizer,
    taxonomy,
    max_length=128
)

test_loader = DataLoader(
    test_ds,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

model = GrieveAIClassifier(
    len(taxonomy["categories"]),
    len(taxonomy["subcategories"])
)

model.to(device)

model.encoder.load_adapter(
    str(Path(CHECKPOINT) / "adapter"),
    adapter_name="default"
)

heads = torch.load(
    str(Path(CHECKPOINT) / "heads.pt"),
    map_location=device
)

model.category_head.load_state_dict(
    heads["category_head"]
)

model.subcategory_head.load_state_dict(
    heads["subcategory_head"]
)

model.priority_head.load_state_dict(
    heads["priority_head"]
)

model.eval()


# ============================================================
# Load TF-IDF + Logistic Regression baseline
# ============================================================

baseline = BaselineGrievanceClassifier.load(
    path="checkpoints/baseline",
    taxonomy_path=TAXONOMY
)

print("Loaded TF-IDF + Logistic Regression baseline.")


# ============================================================
# Build category -> valid subcategories mapping
# ============================================================

category_to_subcategories = {}

for item in taxonomy["categories"]:
    category_to_subcategories[item["name"]] = item["subcategories"]


# ============================================================
# MuRIL category predictions
# ============================================================

muril_categories = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        token_type_ids = batch.get("token_type_ids")

        if token_type_ids is not None:
            token_type_ids = token_type_ids.to(device)

        output = model(
            input_ids,
            attention_mask,
            token_type_ids=token_type_ids
        )

        category_indices = output["category_logits"].argmax(-1)

        muril_categories.extend(
            category_indices.cpu().tolist()
        )


# Convert MuRIL category indices to category names

categories = taxonomy["categories"]
subcategories = taxonomy["subcategories"]

category_names = [
    item["name"]
    for item in categories
]

muril_category_names = [
    category_names[index]
    for index in muril_categories
]


# ============================================================
# Hybrid subcategory prediction
#
# MuRIL decides the category.
# TF-IDF + Logistic Regression decides the subcategory
# within that category.
# ============================================================

hybrid_subcategories = []

for text, predicted_category in zip(
    test_df["text"].tolist(),
    muril_category_names
):

    X = baseline.vectorizer.transform([text])

    sub_probs = baseline.subcategory_clf.predict_proba(X)[0]

    sub_classes = list(
        baseline.subcategory_clf.classes_
    )

    valid_subs = category_to_subcategories[
        predicted_category
    ]

    candidates = []

    for label, probability in zip(
        sub_classes,
        sub_probs
    ):

        if label in valid_subs:

            candidates.append(
                (
                    str(label),
                    float(probability)
                )
            )

    predicted_subcategory = max(
        candidates,
        key=lambda item: item[1]
    )[0]

    hybrid_subcategories.append(
        predicted_subcategory
    )


# ============================================================
# Ground truth
# ============================================================

true_categories = test_df[
    "category"
].tolist()

true_subcategories = test_df[
    "subcategory"
].tolist()


# ============================================================
# Metrics
# ============================================================

category_f1 = f1_score(
    true_categories,
    muril_category_names,
    average="macro"
)

subcategory_f1 = f1_score(
    true_subcategories,
    hybrid_subcategories,
    average="macro"
)


# ============================================================
# Results
# ============================================================

print()
print("=" * 70)
print("EXPERIMENT 4 — MURIL CATEGORY + TF-IDF/LOGREG SUBCATEGORY")
print("=" * 70)

print(
    f"MuRIL category Macro-F1:       "
    f"{category_f1:.4f}"
)

print(
    f"Hybrid subcategory Macro-F1:   "
    f"{subcategory_f1:.4f}"
)

print()
print("REFERENCE RESULTS")
print("-" * 70)

print(
    "TF-IDF/LogReg category Macro-F1:      0.8641"
)

print(
    "TF-IDF/LogReg subcategory Macro-F1:   0.6954"
)

print(
    "MuRIL Run 2 category Macro-F1:        0.9859"
)

print(
    "MuRIL Run 2 subcategory Macro-F1:     0.3046"
)


# ============================================================
# Save predictions
# ============================================================

results = pd.DataFrame({
    "text": test_df["text"].tolist(),
    "true_category": true_categories,
    "muril_category": muril_category_names,
    "true_subcategory": true_subcategories,
    "hybrid_subcategory": hybrid_subcategories
})

results.to_csv(
    "run4_hybrid_predictions.csv",
    index=False
)

print()
print("Saved: run4_hybrid_predictions.csv")
```

```python
%%writefile hybrid_eval.py

from pathlib import Path
import json

import pandas as pd
import torch
from sklearn.metrics import f1_score
from transformers import AutoTokenizer
from torch.utils.data import DataLoader

from src.data_splits import make_splits
from src.model import GrieveAIClassifier, load_taxonomy
from src.train import GrievanceDataset
from src.baseline_classifier import BaselineGrievanceClassifier


DATA = "data/processed/grievances_synthetic.csv"
TAXONOMY = "config/taxonomy.json"
CHECKPOINT = "checkpoints/muril_lora/run2"

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)


# ============================================================
# Load raw taxonomy
# ============================================================

with open(TAXONOMY, "r", encoding="utf-8") as f:
    raw_taxonomy = json.load(f)

category_to_subcategories = raw_taxonomy["categories"]

category_names = list(
    category_to_subcategories.keys()
)

subcategory_names = []

for category in category_names:
    subcategory_names.extend(
        category_to_subcategories[category]["subcategories"]
    )


# ============================================================
# Load model taxonomy
# ============================================================

model_taxonomy = load_taxonomy(TAXONOMY)


# ============================================================
# Load exact same dataset split
# ============================================================

df = pd.read_csv(DATA)

train_df, val_df, test_df = make_splits(
    df,
    seed=42
)

print(
    f"Train: {len(train_df)} | "
    f"Val: {len(val_df)} | "
    f"Test: {len(test_df)}"
)


# ============================================================
# Load MuRIL Run 2
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    str(Path(CHECKPOINT) / "tokenizer")
)

test_ds = GrievanceDataset(
    test_df,
    tokenizer,
    model_taxonomy,
    max_length=128
)

test_loader = DataLoader(
    test_ds,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

model = GrieveAIClassifier(
    len(model_taxonomy["categories"]),
    len(model_taxonomy["subcategories"])
)

model.to(device)

model.encoder.load_adapter(
    str(Path(CHECKPOINT) / "adapter"),
    adapter_name="default"
)

heads = torch.load(
    str(Path(CHECKPOINT) / "heads.pt"),
    map_location=device
)

model.category_head.load_state_dict(
    heads["category_head"]
)

model.subcategory_head.load_state_dict(
    heads["subcategory_head"]
)

model.priority_head.load_state_dict(
    heads["priority_head"]
)

model.eval()


# ============================================================
# Load TF-IDF + Logistic Regression
# ============================================================

baseline = BaselineGrievanceClassifier.load(
    path="checkpoints/baseline",
    taxonomy_path=TAXONOMY
)

print(
    "Loaded TF-IDF + Logistic Regression baseline."
)


# ============================================================
# MuRIL category predictions
# ============================================================

muril_category_indices = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        token_type_ids = batch.get("token_type_ids")

        if token_type_ids is not None:
            token_type_ids = token_type_ids.to(device)

        output = model(
            input_ids,
            attention_mask,
            token_type_ids=token_type_ids
        )

        category_indices = (
            output["category_logits"]
            .argmax(-1)
        )

        muril_category_indices.extend(
            category_indices.cpu().tolist()
        )


# ============================================================
# Convert category indices to names
# ============================================================

muril_category_names = [
    category_names[index]
    for index in muril_category_indices
]


# ============================================================
# Hybrid subcategory prediction
# ============================================================

hybrid_subcategories = []

for text, predicted_category in zip(
    test_df["text"].tolist(),
    muril_category_names
):

    X = baseline.vectorizer.transform(
        [text]
    )

    sub_probs = (
        baseline.subcategory_clf
        .predict_proba(X)[0]
    )

    sub_classes = list(
        baseline.subcategory_clf.classes_
    )

    valid_subcategories = (
        category_to_subcategories[
            predicted_category
        ]["subcategories"]
    )

    candidates = [
        (label, float(probability))
        for label, probability
        in zip(sub_classes, sub_probs)
        if label in valid_subcategories
    ]

    if not candidates:
        raise RuntimeError(
            "No valid subcategory candidates for "
            f"category '{predicted_category}'"
        )

    predicted_subcategory = max(
        candidates,
        key=lambda x: x[1]
    )[0]

    hybrid_subcategories.append(
        predicted_subcategory
    )


# ============================================================
# Ground truth
# ============================================================

true_categories = test_df[
    "category"
].tolist()

true_subcategories = test_df[
    "subcategory"
].tolist()


# ============================================================
# Metrics
# ============================================================

category_f1 = f1_score(
    true_categories,
    muril_category_names,
    average="macro"
)

subcategory_f1 = f1_score(
    true_subcategories,
    hybrid_subcategories,
    average="macro"
)


# ============================================================
# Results
# ============================================================

print()
print("=" * 70)
print(
    "EXPERIMENT 4 — "
    "MURIL CATEGORY + TF-IDF/LOGREG SUBCATEGORY"
)
print("=" * 70)

print(
    f"Hybrid category Macro-F1:       "
    f"{category_f1:.4f}"
)

print(
    f"Hybrid subcategory Macro-F1:    "
    f"{subcategory_f1:.4f}"
)

print()
print("REFERENCE RESULTS")
print("-" * 70)

print("TF-IDF/LogReg category:          0.8641")
print("TF-IDF/LogReg subcategory:       0.6954")
print("MuRIL Run 2 category:            0.9859")
print("MuRIL Run 2 subcategory:         0.3046")


# ============================================================
# Save predictions
# ============================================================

results = pd.DataFrame({
    "text": test_df["text"].tolist(),
    "true_category": true_categories,
    "muril_category": muril_category_names,
    "true_subcategory": true_subcategories,
    "hybrid_subcategory": hybrid_subcategories
})

results.to_csv(
    "run4_hybrid_predictions.csv",
    index=False
)

print()
print("Saved: run4_hybrid_predictions.csv")
```

```python
%%writefile train_indicbert.py

import copy
import json
import random
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from sklearn.metrics import f1_score, mean_absolute_error, recall_score
from torch.utils.data import DataLoader
from transformers import AutoModel, AutoTokenizer
from peft import LoraConfig, TaskType, get_peft_model


# ============================================================
# SETTINGS
# ============================================================

MODEL_NAME = "ai4bharat/IndicBERTv2-MLM-only"

DATA = "data/processed/grievances_synthetic.csv"
TAXONOMY = "config/taxonomy.json"

OUTPUT = "checkpoints/indicbert_lora/run1"

EPOCHS = 10
BATCH_SIZE = 16
LR = 2e-4
MAX_LENGTH = 128
SEED = 42


# ============================================================
# SEED
# ============================================================

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)


# ============================================================
# LOAD TAXONOMY
# ============================================================

with open(TAXONOMY, "r", encoding="utf-8") as f:
    taxonomy = json.load(f)

categories = list(taxonomy["categories"].keys())

subcategories = []

for category in categories:
    subcategories.extend(
        taxonomy["categories"][category]["subcategories"]
    )

cat_to_idx = {
    c: i for i, c in enumerate(categories)
}

sub_to_idx = {
    s: i for i, s in enumerate(subcategories)
}


# ============================================================
# LOAD DATA
# ============================================================

from src.data_splits import make_splits

df = pd.read_csv(DATA)

train_df, val_df, test_df = make_splits(
    df,
    seed=SEED
)

print(
    f"Train: {len(train_df)} | "
    f"Val: {len(val_df)} | "
    f"Test: {len(test_df)}"
)


# ============================================================
# TOKENIZER
# ============================================================

print("Loading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)


# ============================================================
# DATASET
# ============================================================

class IndicDataset(torch.utils.data.Dataset):

    def __init__(self, dataframe):

        self.texts = dataframe["text"].astype(str).tolist()

        self.categories = [
            cat_to_idx[x]
            for x in dataframe["category"]
        ]

        self.subcategories = [
            sub_to_idx[x]
            for x in dataframe["subcategory"]
        ]

        self.priorities = (
            dataframe["priority"]
            .astype(float)
            .tolist()
        )

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):

        encoded = tokenizer(
            self.texts[idx],
            truncation=True,
            padding="max_length",
            max_length=MAX_LENGTH,
            return_tensors="pt"
        )

        item = {
            "input_ids": encoded["input_ids"].squeeze(0),
            "attention_mask": encoded["attention_mask"].squeeze(0),
            "category_label": torch.tensor(
                self.categories[idx],
                dtype=torch.long
            ),
            "subcategory_label": torch.tensor(
                self.subcategories[idx],
                dtype=torch.long
            ),
            "priority_label": torch.tensor(
                self.priorities[idx],
                dtype=torch.float
            )
        }

        if "token_type_ids" in encoded:
            item["token_type_ids"] = (
                encoded["token_type_ids"].squeeze(0)
            )

        return item


train_ds = IndicDataset(train_df)
val_ds = IndicDataset(val_df)
test_ds = IndicDataset(test_df)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE
)


# ============================================================
# MODEL
# ============================================================

print("Loading IndicBERT v2...")

encoder = AutoModel.from_pretrained(
    MODEL_NAME
)

encoder.to(device)


# ============================================================
# LoRA
# ============================================================

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    lora_dropout=0.1,
    target_modules=["query", "value"],
    task_type=TaskType.FEATURE_EXTRACTION,
    bias="none"
)

encoder = get_peft_model(
    encoder,
    lora_config
)

encoder.print_trainable_parameters()


# ============================================================
# CLASSIFICATION HEADS
# ============================================================

hidden_size = encoder.config.hidden_size

category_head = torch.nn.Linear(
    hidden_size,
    len(categories)
).to(device)

subcategory_head = torch.nn.Linear(
    hidden_size,
    len(subcategories)
).to(device)

priority_head = torch.nn.Linear(
    hidden_size,
    1
).to(device)


# ============================================================
# POOLING
# ============================================================

def mean_pool(
    hidden_states,
    attention_mask
):

    mask = attention_mask.unsqueeze(-1).float()

    summed = (
        hidden_states * mask
    ).sum(dim=1)

    counts = mask.sum(dim=1).clamp(
        min=1e-9
    )

    return summed / counts


# ============================================================
# CATEGORY-SUBCATEGORY MASK
# ============================================================

subcategory_to_category = {}

for category in categories:

    for subcategory in taxonomy[
        "categories"
    ][category]["subcategories"]:

        subcategory_to_category[
            subcategory
        ] = category


category_to_valid_sub_indices = {}

for category in categories:

    valid = taxonomy[
        "categories"
    ][category]["subcategories"]

    category_to_valid_sub_indices[
        cat_to_idx[category]
    ] = [
        sub_to_idx[s]
        for s in valid
    ]


def mask_subcategory_logits(
    logits,
    category_index
):

    valid_indices = category_to_valid_sub_indices[
        int(category_index)
    ]

    masked = torch.full_like(
        logits,
        -1e9
    )

    masked[
        valid_indices
    ] = logits[
        valid_indices
    ]

    return masked


# ============================================================
# FORWARD
# ============================================================

def forward(
    input_ids,
    attention_mask,
    token_type_ids=None
):

    kwargs = {
        "input_ids": input_ids,
        "attention_mask": attention_mask
    }

    if token_type_ids is not None:
        kwargs["token_type_ids"] = token_type_ids

    output = encoder(**kwargs)

    pooled = mean_pool(
        output.last_hidden_state,
        attention_mask
    )

    return (
        category_head(pooled),
        subcategory_head(pooled),
        priority_head(pooled).squeeze(-1)
    )


# ============================================================
# OPTIMIZER
# ============================================================

trainable_params = (
    list(encoder.parameters())
    + list(category_head.parameters())
    + list(subcategory_head.parameters())
    + list(priority_head.parameters())
)

optimizer = torch.optim.AdamW(
    [
        p for p in trainable_params
        if p.requires_grad
    ],
    lr=LR,
    weight_decay=0.01
)

cat_loss_fn = torch.nn.CrossEntropyLoss()
sub_loss_fn = torch.nn.CrossEntropyLoss()
priority_loss_fn = torch.nn.HuberLoss()


# ============================================================
# EVALUATION
# ============================================================

def evaluate(loader):

    encoder.eval()
    category_head.eval()
    subcategory_head.eval()
    priority_head.eval()

    y_cat = []
    p_cat = []

    y_sub = []
    p_sub = []

    y_sub_oracle = []
    p_sub_oracle = []

    y_pri = []
    p_pri = []

    high_true = []
    high_pred = []

    with torch.no_grad():

        for batch in loader:

            input_ids = batch[
                "input_ids"
            ].to(device)

            attention_mask = batch[
                "attention_mask"
            ].to(device)

            token_type_ids = batch.get(
                "token_type_ids"
            )

            if token_type_ids is not None:
                token_type_ids = token_type_ids.to(device)

            cat_logits, sub_logits, pri = forward(
                input_ids,
                attention_mask,
                token_type_ids
            )

            cat_pred = cat_logits.argmax(-1)

            masked_sub = torch.stack([
                mask_subcategory_logits(
                    sub_logits[i],
                    int(cat_pred[i])
                )
                for i in range(len(cat_pred))
            ])

            sub_pred = masked_sub.argmax(-1)

            oracle_sub = torch.stack([
                mask_subcategory_logits(
                    sub_logits[i],
                    int(
                        batch[
                            "category_label"
                        ][i]
                    )
                )
                for i in range(
                    len(
                        batch[
                            "category_label"
                        ]
                    )
                )
            ]).argmax(-1)

            y_cat.extend(
                batch[
                    "category_label"
                ].cpu().tolist()
            )

            p_cat.extend(
                cat_pred.cpu().tolist()
            )

            y_sub.extend(
                batch[
                    "subcategory_label"
                ].cpu().tolist()
            )

            p_sub.extend(
                sub_pred.cpu().tolist()
            )

            y_sub_oracle.extend(
                batch[
                    "subcategory_label"
                ].cpu().tolist()
            )

            p_sub_oracle.extend(
                oracle_sub.cpu().tolist()
            )

            y_pri.extend(
                batch[
                    "priority_label"
                ].cpu().tolist()
            )

            p_pri.extend(
                pri.cpu().tolist()
            )

            high_true.extend(
                (
                    batch[
                        "priority_label"
                    ] >= 4
                ).int().tolist()
            )

            high_pred.extend(
                (pri >= 4).int().tolist()
            )

    return {
        "category_macro_f1": f1_score(
            y_cat,
            p_cat,
            average="macro"
        ),

        "subcategory_macro_f1": f1_score(
            y_sub,
            p_sub,
            average="macro"
        ),

        "subcategory_macro_f1_given_gold_category":
            f1_score(
                y_sub_oracle,
                p_sub_oracle,
                average="macro"
            ),

        "priority_mae": mean_absolute_error(
            y_pri,
            p_pri
        ),

        "priority_recall_high": recall_score(
            high_true,
            high_pred,
            zero_division=0
        )
    }


# ============================================================
# TRAIN
# ============================================================

best_score = -1
best_state = None

for epoch in range(
    1,
    EPOCHS + 1
):

    encoder.train()
    category_head.train()
    subcategory_head.train()
    priority_head.train()

    total_loss = 0

    for batch in train_loader:

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        token_type_ids = batch.get(
            "token_type_ids"
        )

        if token_type_ids is not None:
            token_type_ids = token_type_ids.to(device)

        cat_labels = batch[
            "category_label"
        ].to(device)

        sub_labels = batch[
            "subcategory_label"
        ].to(device)

        pri_labels = batch[
            "priority_label"
        ].to(device)

        optimizer.zero_grad()

        cat_logits, sub_logits, pri = forward(
            input_ids,
            attention_mask,
            token_type_ids
        )

        conditional_sub_logits = torch.stack([
            mask_subcategory_logits(
                sub_logits[i],
                int(cat_labels[i])
            )
            for i in range(
                len(cat_labels)
            )
        ])

        loss = (
            cat_loss_fn(
                cat_logits,
                cat_labels
            )
            +
            sub_loss_fn(
                conditional_sub_logits,
                sub_labels
            )
            +
            0.1 * priority_loss_fn(
                pri,
                pri_labels
            )
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            1.0
        )

        optimizer.step()

        total_loss += float(
            loss.item()
        )

    metrics = evaluate(val_loader)

    score = (
        metrics["category_macro_f1"]
        +
        metrics["subcategory_macro_f1"]
    )

    if score > best_score:

        best_score = score

        best_state = {
            "encoder": copy.deepcopy(
                encoder.state_dict()
            ),

            "category_head": copy.deepcopy(
                category_head.state_dict()
            ),

            "subcategory_head": copy.deepcopy(
                subcategory_head.state_dict()
            ),

            "priority_head": copy.deepcopy(
                priority_head.state_dict()
            )
        }

    print(
        f"Epoch {epoch}/{EPOCHS} "
        f"loss={total_loss / len(train_loader):.4f} "
        f"metrics={metrics}"
    )


# ============================================================
# BEST CHECKPOINT → TEST
# ============================================================

encoder.load_state_dict(
    best_state["encoder"]
)

category_head.load_state_dict(
    best_state["category_head"]
)

subcategory_head.load_state_dict(
    best_state["subcategory_head"]
)

priority_head.load_state_dict(
    best_state["priority_head"]
)

test_metrics = evaluate(
    test_loader
)

print()
print("=" * 70)
print("INDICBERT V2 — FINAL TEST")
print("=" * 70)

print(
    json.dumps(
        test_metrics,
        indent=2
    )
)


# ============================================================
# SAVE
# ============================================================

output_dir = Path(OUTPUT)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

encoder.save_pretrained(
    output_dir / "adapter"
)

tokenizer.save_pretrained(
    output_dir / "tokenizer"
)

torch.save(
    {
        "category_head":
            category_head.state_dict(),

        "subcategory_head":
            subcategory_head.state_dict(),

        "priority_head":
            priority_head.state_dict()
    },
    output_dir / "heads.pt"
)

Path(
    output_dir / "taxonomy.json"
).write_text(
    json.dumps(
        taxonomy,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

Path(
    output_dir / "metrics.json"
).write_text(
    json.dumps(
        test_metrics,
        indent=2
    ),
    encoding="utf-8"
)

print()
print(
    f"Checkpoint saved to {OUTPUT}"
)
```

```python
%%writefile train_indicbert_60k.py

import json
import random
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import f1_score, mean_absolute_error, recall_score
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel
from peft import LoraConfig, TaskType, get_peft_model


# ============================================================
# CONFIG
# ============================================================

MODEL_NAME = "ai4bharat/IndicBERTv2-MLM-only"

DATA_PATH = "/content/grieveai_synthetic_60000_stress.csv"

OUTPUT_DIR = Path("checkpoints/indicbert_lora/stress_60k")

SEED = 42

# 60k stress test:
# 48,000 train / 6,000 validation / 6,000 test
EPOCHS = 3
BATCH_SIZE = 64

LR = 2e-4
MAX_LENGTH = 128


# ============================================================
# TAXONOMY
# ============================================================

TAXONOMY = {
    "Academics": [
        "timetable_scheduling",
        "teaching_quality",
        "faculty_conduct",
        "syllabus_course_progress",
        "attendance",
    ],
    "Examinations": [
        "hall_ticket",
        "marks_discrepancy",
        "revaluation",
        "examination_schedule",
        "examination_process",
    ],
    "Fees_Accounts": [
        "fee_discrepancy",
        "refund",
        "scholarship",
        "payment_transaction",
        "accounts_financial_documentation",
    ],
    "IT_Library": [
        "portal_account_access",
        "wifi_network",
        "software_license",
        "library_book_availability",
        "library_digital_resources",
    ],
    "Infrastructure": [
        "classroom_lab_maintenance",
        "electrical_issues",
        "sanitation_cleanliness",
        "parking_access",
        "construction_facility_disruption",
    ],
    "Transport": [
        "commuting_accessibility",
        "transport_academic_conflict",
        "travel_safety_access",
    ],
    "Canteen": [
        "food_quality",
        "food_hygiene",
        "pricing",
        "menu_variety",
        "food_service",
    ],
}

CATEGORIES = list(TAXONOMY.keys())

CATEGORY_TO_ID = {
    x: i for i, x in enumerate(CATEGORIES)
}

SUBCATEGORIES = []

SUBCATEGORY_TO_ID = {}

SUBCATEGORY_TO_CATEGORY = {}

for category, subs in TAXONOMY.items():

    for sub in subs:

        SUBCATEGORY_TO_ID[sub] = len(
            SUBCATEGORIES
        )

        SUBCATEGORIES.append(sub)

        SUBCATEGORY_TO_CATEGORY[sub] = category


# ============================================================
# REPRODUCIBILITY
# ============================================================

def seed_everything(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    torch.cuda.manual_seed_all(seed)


# ============================================================
# DATA SPLIT
# ============================================================

def split_data(df):

    df = df.sample(
        frac=1,
        random_state=SEED
    ).reset_index(drop=True)

    n = len(df)

    train_end = int(n * 0.80)

    val_end = int(n * 0.90)

    train_df = df.iloc[
        :train_end
    ].reset_index(drop=True)

    val_df = df.iloc[
        train_end:val_end
    ].reset_index(drop=True)

    test_df = df.iloc[
        val_end:
    ].reset_index(drop=True)

    return train_df, val_df, test_df


# ============================================================
# DATASET
# ============================================================

class GrievanceDataset(Dataset):

    def __init__(self, df, tokenizer):

        self.df = df.reset_index(drop=True)

        self.tokenizer = tokenizer

    def __len__(self):

        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        enc = self.tokenizer(
            str(row["text"]),
            truncation=True,
            padding="max_length",
            max_length=MAX_LENGTH,
            return_tensors="pt",
        )

        category_id = CATEGORY_TO_ID[
            row["category"]
        ]

        subcategory_id = SUBCATEGORY_TO_ID[
            row["subcategory"]
        ]

        item = {

            "input_ids":
                enc["input_ids"].squeeze(0),

            "attention_mask":
                enc["attention_mask"].squeeze(0),

            "category_label":
                torch.tensor(
                    category_id,
                    dtype=torch.long
                ),

            "subcategory_label":
                torch.tensor(
                    subcategory_id,
                    dtype=torch.long
                ),

            "priority_label":
                torch.tensor(
                    float(row["priority"]),
                    dtype=torch.float
                ),
        }

        if "token_type_ids" in enc:

            item["token_type_ids"] = (
                enc["token_type_ids"]
                .squeeze(0)
            )

        return item


# ============================================================
# MODEL
# ============================================================

class GrieveAIIndicBERT(nn.Module):

    def __init__(self):

        super().__init__()

        base = AutoModel.from_pretrained(
            MODEL_NAME
        )

        lora_config = LoraConfig(
            r=8,
            lora_alpha=16,
            lora_dropout=0.1,
            target_modules=[
                "query",
                "value"
            ],
            bias="none",
            task_type=TaskType.FEATURE_EXTRACTION,
        )

        self.encoder = get_peft_model(
            base,
            lora_config
        )

        hidden_size = base.config.hidden_size

        self.category_head = nn.Linear(
            hidden_size,
            len(CATEGORIES)
        )

        self.subcategory_head = nn.Linear(
            hidden_size,
            len(SUBCATEGORIES)
        )

        self.priority_head = nn.Linear(
            hidden_size,
            1
        )

        self.dropout = nn.Dropout(0.1)

    def mean_pool(self, hidden, mask):

        mask = mask.unsqueeze(-1).float()

        summed = (
            hidden * mask
        ).sum(dim=1)

        denom = (
            mask.sum(dim=1)
            .clamp(min=1e-9)
        )

        return summed / denom

    def forward(
        self,
        input_ids,
        attention_mask,
        token_type_ids=None,
    ):

        kwargs = {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
        }

        if token_type_ids is not None:

            kwargs["token_type_ids"] = (
                token_type_ids
            )

        outputs = self.encoder(
            **kwargs
        )

        pooled = self.mean_pool(
            outputs.last_hidden_state,
            attention_mask
        )

        pooled = self.dropout(
            pooled
        )

        return {

            "category_logits":
                self.category_head(
                    pooled
                ),

            "subcategory_logits":
                self.subcategory_head(
                    pooled
                ),

            "priority_pred":
                self.priority_head(
                    pooled
                ).squeeze(-1),
        }

    def mask_subcategories(
        self,
        logits,
        category_id
    ):

        category = CATEGORIES[
            category_id
        ]

        valid_subs = TAXONOMY[
            category
        ]

        masked = torch.full_like(
            logits,
            -1e9
        )

        for sub in valid_subs:

            sid = SUBCATEGORY_TO_ID[
                sub
            ]

            masked[sid] = logits[sid]

        return masked


# ============================================================
# EVALUATION
# ============================================================

def evaluate(
    model,
    loader,
    device
):

    model.eval()

    y_cat = []
    p_cat = []

    y_sub = []
    p_sub = []

    y_sub_oracle = []
    p_sub_oracle = []

    y_pri = []
    p_pri = []

    high_true = []
    high_pred = []

    with torch.no_grad():

        for batch in loader:

            input_ids = batch[
                "input_ids"
            ].to(device)

            attention_mask = batch[
                "attention_mask"
            ].to(device)

            token_type_ids = batch.get(
                "token_type_ids"
            )

            if token_type_ids is not None:

                token_type_ids = (
                    token_type_ids.to(device)
                )

            out = model(
                input_ids,
                attention_mask,
                token_type_ids,
            )

            cat = (
                out["category_logits"]
                .argmax(-1)
            )

            predicted_sub = []

            oracle_sub = []

            for i in range(len(cat)):

                masked_pred = (
                    model.mask_subcategories(
                        out[
                            "subcategory_logits"
                        ][i],
                        int(cat[i])
                    )
                )

                masked_gold = (
                    model.mask_subcategories(
                        out[
                            "subcategory_logits"
                        ][i],
                        int(
                            batch[
                                "category_label"
                            ][i]
                        )
                    )
                )

                predicted_sub.append(
                    masked_pred.argmax()
                )

                oracle_sub.append(
                    masked_gold.argmax()
                )

            sub = torch.stack(
                predicted_sub
            )

            oracle = torch.stack(
                oracle_sub
            )

            pri = out[
                "priority_pred"
            ]

            y_cat.extend(
                batch[
                    "category_label"
                ].tolist()
            )

            p_cat.extend(
                cat.cpu().tolist()
            )

            y_sub.extend(
                batch[
                    "subcategory_label"
                ].tolist()
            )

            p_sub.extend(
                sub.cpu().tolist()
            )

            y_sub_oracle.extend(
                batch[
                    "subcategory_label"
                ].tolist()
            )

            p_sub_oracle.extend(
                oracle.cpu().tolist()
            )

            y_pri.extend(
                batch[
                    "priority_label"
                ].tolist()
            )

            p_pri.extend(
                pri.cpu().tolist()
            )

            high_true.extend(
                (
                    batch[
                        "priority_label"
                    ] >= 4
                )
                .int()
                .tolist()
            )

            high_pred.extend(
                (pri >= 4)
                .int()
                .tolist()
            )

    return {

        "category_macro_f1":
            f1_score(
                y_cat,
                p_cat,
                average="macro"
            ),

        "subcategory_macro_f1":
            f1_score(
                y_sub,
                p_sub,
                average="macro"
            ),

        "subcategory_macro_f1_given_gold_category":
            f1_score(
                y_sub_oracle,
                p_sub_oracle,
                average="macro"
            ),

        "priority_mae":
            mean_absolute_error(
                y_pri,
                p_pri
            ),

        "priority_recall_high":
            recall_score(
                high_true,
                high_pred,
                zero_division=0
            ),
    }


# ============================================================
# MAIN
# ============================================================

def main():

    seed_everything(SEED)

    device = torch.device(
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

    print(
        "Device:",
        device
    )

    df = pd.read_csv(
        DATA_PATH
    )

    print(
        f"Dataset: {len(df):,} records"
    )

    print(
        "Categories:",
        df["category"].nunique()
    )

    print(
        "Subcategories:",
        df["subcategory"].nunique()
    )

    train_df, val_df, test_df = (
        split_data(df)
    )

    print()

    print(
        f"Train: {len(train_df):,}"
    )

    print(
        f"Validation: {len(val_df):,}"
    )

    print(
        f"Test: {len(test_df):,}"
    )

    tokenizer = (
        AutoTokenizer.from_pretrained(
            MODEL_NAME
        )
    )

    train_ds = GrievanceDataset(
        train_df,
        tokenizer
    )

    val_ds = GrievanceDataset(
        val_df,
        tokenizer
    )

    test_ds = GrievanceDataset(
        test_df,
        tokenizer
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=0,
        pin_memory=True,
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=BATCH_SIZE,
        num_workers=0,
        pin_memory=True,
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=BATCH_SIZE,
        num_workers=0,
        pin_memory=True,
    )

    print()

    print(
        "Loading IndicBERT v2..."
    )

    model = GrieveAIIndicBERT()

    model.to(device)

    model.encoder.print_trainable_parameters()

    ce_loss = nn.CrossEntropyLoss()

    huber_loss = nn.HuberLoss()

    optimizer = torch.optim.AdamW(
        filter(
            lambda p:
                p.requires_grad,
            model.parameters()
        ),
        lr=LR,
        weight_decay=0.01,
    )

    best_score = -1

    OUTPUT_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    for epoch in range(
        1,
        EPOCHS + 1
    ):

        model.train()

        total_loss = 0

        for batch in train_loader:

            optimizer.zero_grad()

            input_ids = batch[
                "input_ids"
            ].to(device)

            attention_mask = batch[
                "attention_mask"
            ].to(device)

            token_type_ids = batch.get(
                "token_type_ids"
            )

            if token_type_ids is not None:

                token_type_ids = (
                    token_type_ids.to(device)
                )

            cat_labels = batch[
                "category_label"
            ].to(device)

            sub_labels = batch[
                "subcategory_label"
            ].to(device)

            pri_labels = batch[
                "priority_label"
            ].to(device)

            out = model(
                input_ids,
                attention_mask,
                token_type_ids,
            )

            cat_loss = ce_loss(
                out["category_logits"],
                cat_labels
            )

            # Hierarchical subcategory loss.
            # Restrict each example to
            # its gold category.

            sub_losses = []

            for i in range(
                len(cat_labels)
            ):

                masked = (
                    model.mask_subcategories(
                        out[
                            "subcategory_logits"
                        ][i],
                        int(
                            cat_labels[i]
                        )
                    )
                )

                sub_losses.append(
                    ce_loss(
                        masked.unsqueeze(0),
                        sub_labels[
                            i
                        ].unsqueeze(0)
                    )
                )

            sub_loss = torch.stack(
                sub_losses
            ).mean()

            pri_loss = huber_loss(
                out["priority_pred"],
                pri_labels
            )

            loss = (
                cat_loss
                + sub_loss
                + 0.1 * pri_loss
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                1.0
            )

            optimizer.step()

            total_loss += (
                loss.item()
            )

        metrics = evaluate(
            model,
            val_loader,
            device
        )

        avg_loss = (
            total_loss
            / len(train_loader)
        )

        print(
            f"Epoch {epoch}/{EPOCHS} "
            f"loss={avg_loss:.4f} "
            f"metrics={metrics}"
        )

        score = (
            metrics[
                "category_macro_f1"
            ]
            +
            metrics[
                "subcategory_macro_f1"
            ]
        )

        if score > best_score:

            best_score = score

            torch.save(
                {
                    "model_state_dict":
                        model.state_dict(),

                    "epoch":
                        epoch,

                    "metrics":
                        metrics,
                },
                OUTPUT_DIR
                / "best_model.pt"
            )

    # ========================================================
    # FINAL TEST
    # ========================================================

    checkpoint = torch.load(
        OUTPUT_DIR
        / "best_model.pt",
        map_location=device
    )

    model.load_state_dict(
        checkpoint[
            "model_state_dict"
        ]
    )

    test_metrics = evaluate(
        model,
        test_loader,
        device
    )

    print()

    print("=" * 70)

    print(
        "INDICBERT V2 — 60K STRESS TEST"
    )

    print("=" * 70)

    print(
        json.dumps(
            test_metrics,
            indent=2
        )
    )

    # ========================================================
    # SAVE EXPERIMENT METADATA
    # ========================================================

    with open(
        OUTPUT_DIR
        / "metrics.json",
        "w"
    ) as f:

        json.dump(
            {
                "dataset":
                    "synthetic_60000_stress",

                "train_records":
                    len(train_df),

                "validation_records":
                    len(val_df),

                "test_records":
                    len(test_df),

                "best_epoch":
                    checkpoint["epoch"],

                "test_metrics":
                    test_metrics,

                "seed":
                    SEED,

                "epochs":
                    EPOCHS,

                "batch_size":
                    BATCH_SIZE,

                "learning_rate":
                    LR,

                "max_length":
                    MAX_LENGTH,
            },
            f,
            indent=2
        )

    print()

    print(
        "Checkpoint saved to:",
        OUTPUT_DIR
    )


if __name__ == "__main__":

    main()
```
